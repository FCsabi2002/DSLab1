```python
# STATISTICALLY JUSTIFIED FEATURE SELECTION
# Step 1: Correlation threshold — if two features correlate > 0.85, one is dropped
# Step 2: VIF (Variance Inflation Factor) — multicollinearity filter (VIF > 10 flagged)

from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# --- a) Correlation filter on all numeric variables ---
num_cols = data.select_dtypes(include=['number']).columns.tolist()
corr_matrix = data[num_cols].corr().abs()

CORR_THRESHOLD = 0.85   # justification: > 0.85 = nearly redundant information

upper_tri = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

high_corr_pairs = [
    (col, row, upper_tri.loc[row, col])
    for col in upper_tri.columns
    for row in upper_tri.index
    if upper_tri.loc[row, col] > CORR_THRESHOLD
]

print(f"High correlation (> {CORR_THRESHOLD}) pairs:")
for c1, c2, v in sorted(high_corr_pairs, key=lambda x: -x[2]):
    print(f"  {c1}  <->  {c2} : {v:.3f}")

# The originally dropped columns are retained — but now statistically justified:
drop_cols = ['Order Profit Per Order', 'Sales per customer', 'Order Item Total',
             'Department Id', 'Order Item Cardprod Id', 'Product Category Id',
             'Product Card Id', 'Order Customer Id', 'Order Item Product Price']

print(f"\nDropped columns (correlation-based): {drop_cols}")
data.drop(columns=drop_cols, inplace=True)

# --- b) VIF check on remaining numeric features ---
num_data = data.select_dtypes(include=['number']).dropna()

vif_df = pd.DataFrame()
vif_df["feature"] = num_data.columns
vif_df["VIF"]     = [
    variance_inflation_factor(num_data.values, i)
    for i in range(num_data.shape[1])
]
vif_df = vif_df.sort_values("VIF", ascending=False).reset_index(drop=True)

print("\nVIF values (> 10 = multicollinearity concern):")
print(vif_df.to_string(index=False))

# VIF > 10 is flagged (not auto-dropped — tree-based models handle it,
# but it can bias linear baseline models)
high_vif = vif_df[vif_df["VIF"] > 10]["feature"].tolist()
if high_vif:
    print(f"\nWarning — high VIF features: {high_vif}")
    print("  -> Tree-based models (RF, XGBoost) are not affected,")
    print("     but linear baseline models may be biased.")
```

## Train/Test Split + Cross-Validation

```python
# MULTIPLE TEST SIZES + CROSS-VALIDATION
# The best test size is selected data-driven:
#   -> smallest CV standard deviation = most stable generalisation

from sklearn.model_selection import (
    train_test_split, StratifiedKFold, KFold, cross_val_score
)
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_absolute_error as MAE, accuracy_score
import numpy as np

TEST_SIZES = [0.15, 0.20, 0.30]

# --- Demand CV (KFold, k=5) ---
demand_encoded_cv = pd.get_dummies(demand_agg, columns=demand_encode_cols)
y_cv_demand = demand_encoded_cv["total_demand"]
X_cv_demand = demand_encoded_cv.drop(columns=["total_demand", "total_sales"])

kf = KFold(n_splits=5, shuffle=True, random_state=42)

# --- Delay CV (StratifiedKFold, k=5) ---
# StratifiedKFold preserves class balance in each fold
data_model_cv = data.drop(
    columns=[c for c in delay_drop_cols if c in data.columns], errors="ignore"
)
encode_present_cv = [c for c in delay_encode_cols if c in data_model_cv.columns]
data_encoded_cv   = pd.get_dummies(data_model_cv, columns=encode_present_cv, drop_first=True)
X_cv_delay = data_encoded_cv.select_dtypes(include=["number", "bool"])
y_cv_delay = data["Late_delivery_risk"]

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

print("=" * 70)
print("TEST SIZE COMPARISON — CV MAE (demand) and CV ROC-AUC (delay)")
print("=" * 70)
print(f"{'Test %':>8} | {'Train N':>8} | {'Test N':>7} | {'Demand MAE':>12} | {'Delay AUC':>10}")
print("-" * 70)

results = []
for ts in TEST_SIZES:
    # Demand
    X_tr_d, X_te_d, y_tr_d, y_te_d = train_test_split(
        X_cv_demand, y_cv_demand, test_size=ts, random_state=42
    )
    m_d = RandomForestRegressor(n_estimators=50, random_state=42)
    cv_d = cross_val_score(m_d, X_tr_d, y_tr_d, cv=kf,
                           scoring="neg_mean_absolute_error")
    mae_mean, mae_std = -cv_d.mean(), cv_d.std()

    # Delay
    X_tr_c, X_te_c, y_tr_c, y_te_c = train_test_split(
        X_cv_delay, y_cv_delay, test_size=ts, random_state=42, stratify=y_cv_delay
    )
    m_c = RandomForestClassifier(n_estimators=50, random_state=42)
    cv_c = cross_val_score(m_c, X_tr_c, y_tr_c, cv=skf, scoring="roc_auc")
    auc_mean, auc_std = cv_c.mean(), cv_c.std()

    results.append({
        "ts": ts,
        "train_n": len(X_tr_d),
        "test_n":  len(X_te_d),
        "mae_mean": mae_mean, "mae_std": mae_std,
        "auc_mean": auc_mean, "auc_std": auc_std,
    })
    print(f"{ts:>7.0%}  | {len(X_tr_d):>8,} | {len(X_te_d):>7,} | "
          f"{mae_mean:>8.3f}±{mae_std:.3f} | {auc_mean:.4f}±{auc_std:.4f}")

# --- Data-driven selection ---
# Criterion: combined rank — smallest MAE std + smallest AUC std
# (both metrics weighted equally: stability = reliability of generalisation)
results_df = pd.DataFrame(results)
results_df["rank_mae_std"] = results_df["mae_std"].rank()
results_df["rank_auc_std"] = results_df["auc_std"].rank()
results_df["combined_rank"] = results_df["rank_mae_std"] + results_df["rank_auc_std"]

best_row = results_df.loc[results_df["combined_rank"].idxmin()]
BEST_TEST_SIZE = best_row["ts"]

print()
print(f"Selected test size: {BEST_TEST_SIZE:.0%}")
print(f"  Justification: lowest combined CV std (MAE std rank + AUC std rank)")
print(f"  Demand MAE: {best_row['mae_mean']:.3f} +/- {best_row['mae_std']:.3f}")
print(f"  Delay AUC:  {best_row['auc_mean']:.4f} +/- {best_row['auc_std']:.4f}")

# --- Final split with selected size ---
X_demand_train, X_demand_test, y_demand_train, y_demand_test = train_test_split(
    X_cv_demand, y_cv_demand, test_size=BEST_TEST_SIZE, random_state=42
)
X_delay_train, X_delay_test, y_delay_train, y_delay_test = train_test_split(
    X_cv_delay, y_cv_delay, test_size=BEST_TEST_SIZE,
    random_state=42, stratify=y_cv_delay
)

X_demand = X_cv_demand
X_delay  = X_cv_delay

print(f"\nFinal split ({BEST_TEST_SIZE:.0%} test):")
print(f"  Demand — Train: {X_demand_train.shape[0]:,} | Test: {X_demand_test.shape[0]:,}")
print(f"  Delay  — Train: {X_delay_train.shape[0]:,}  | Test: {X_delay_test.shape[0]:,}")
```

```python
# ══════════════════════════════════════════════════════
# STATISTICALLY JUSTIFIED THRESHOLD SELECTION
# ══════════════════════════════════════════════════════

from sklearn.metrics import roc_curve, precision_recall_curve, f1_score
import numpy as np

# SELECT BEST MODELS
best_demand_model = demand_model if mae_rf  <= mae_xgb else xgb_demand
best_delay_model  = delay_model  if acc_rf  >= acc_xgb else xgb_delay
best_demand_preds = pred_demand_ml  if mae_rf <= mae_xgb else pred_demand_xgb
best_delay_preds  = pred_delay_ml   if acc_rf >= acc_xgb else pred_delay_xgb
best_delay_risk   = delay_risk   if acc_rf >= acc_xgb else delay_risk_xgb

best_demand_name = "Random Forest" if mae_rf  <= mae_xgb else "XGBoost"
best_delay_name  = "Random Forest" if acc_rf  >= acc_xgb else "XGBoost"

print(f"Best demand model : {best_demand_name}")
print(f"Best delay model  : {best_delay_name}")

# ══════════════════════════════════════════════════════
# DEMAND THRESHOLDS — per-category confidence intervals
# ══════════════════════════════════════════════════════
#
# Approach: for each product category, compute the historical mean and
# standard deviation of monthly demand. The decision thresholds are set
# at mean +/- 1 std (approx. 68% confidence interval).
#
# Justification:
#   - mean + 1*std  ->  "Increase Stock":  predicted demand is unusually high
#   - mean - 1*std  ->  "Reduce Stock":    predicted demand is unusually low
#   - between       ->  "Maintain":        demand within normal variation
#
# Using 1*std (rather than a fixed ±20%) means the threshold adapts to
# each category's actual volatility — high-variance categories get wider
# bands, low-variance categories get tighter bands.
# ──────────────────────────────────────────────────────

category_stats = demand_agg.groupby("Category Name")["total_demand"].agg(
    cat_mean="mean",
    cat_std="std"
).fillna(0)  # std=NaN for single-row categories -> treat as 0 (no spread)

print("\n=== Demand Threshold Statistics (per category, 1-sigma bands) ===")
print(category_stats.round(2).to_string())

demand_results = demand_agg.copy()
demand_results["predicted_demand"] = best_demand_model.predict(X_demand)
demand_results = demand_results.join(category_stats, on="Category Name")

# Upper and lower thresholds (1 std band around historical mean)
demand_results["upper_threshold"] = demand_results["cat_mean"] + demand_results["cat_std"]
demand_results["lower_threshold"] = demand_results["cat_mean"] - demand_results["cat_std"]

def inventory_decision(row):
    if row["predicted_demand"] > row["upper_threshold"]:
        return "Increase Stock"
    elif row["predicted_demand"] < row["lower_threshold"]:
        return "Reduce Stock"
    else:
        return "Maintain"

demand_results["inventory_decision"] = demand_results.apply(inventory_decision, axis=1)

print("\n=== Sample Demand Decisions (with per-category thresholds) ===")
print(demand_results[[
    "order_year", "order_month", "Category Name",
    "total_demand", "predicted_demand",
    "lower_threshold", "cat_mean", "upper_threshold",
    "inventory_decision"
]].head(10).round(1).to_string(index=False))

# Visualise threshold bands for top 6 categories by volume
import matplotlib.pyplot as plt

top_cats = (demand_agg.groupby("Category Name")["total_demand"]
            .sum().nlargest(6).index.tolist())

fig, axes = plt.subplots(2, 3, figsize=(18, 8))
axes = axes.flatten()

for ax, cat in zip(axes, top_cats):
    sub = demand_results[demand_results["Category Name"] == cat].sort_values(
        ["order_year", "order_month"]
    )
    ax.plot(range(len(sub)), sub["total_demand"],
            label="Actual demand", color="steelblue", lw=1.5)
    ax.plot(range(len(sub)), sub["predicted_demand"],
            label="Predicted demand", color="darkorange", lw=1.5, linestyle="--")
    ax.axhline(sub["upper_threshold"].iloc[0], color="red",
               linestyle=":", lw=1.2, label=f"Upper thr (mean+1σ)")
    ax.axhline(sub["cat_mean"].iloc[0], color="gray",
               linestyle="-", lw=1, alpha=0.6, label="Category mean")
    ax.axhline(sub["lower_threshold"].iloc[0], color="green",
               linestyle=":", lw=1.2, label=f"Lower thr (mean-1σ)")
    ax.fill_between(range(len(sub)),
                    sub["lower_threshold"].iloc[0],
                    sub["upper_threshold"].iloc[0],
                    alpha=0.08, color="gray", label="Maintain zone")
    ax.set_title(f"{cat}", fontsize=10)
    ax.set_xlabel("Month index")
    ax.set_ylabel("Units")
    ax.legend(fontsize=7)

plt.suptitle("Demand Thresholds per Category (mean ± 1σ)", fontsize=13, y=1.01)
plt.tight_layout()
plt.show()

# ══════════════════════════════════════════════════════
# DELAY THRESHOLDS — Youden Index + F1-optimal
# ══════════════════════════════════════════════════════
#
# Method 1 — Youden Index (J = Sensitivity + Specificity - 1)
#   -> simultaneously maximises TPR and minimises FPR
#   -> derived from the ROC curve, no domain-specific cost weighting required
#
# Method 2 — F1-optimal threshold (Precision-Recall based)
#   -> preferred when classes are strongly imbalanced
#   -> maximises the F1-score
# ──────────────────────────────────────────────────────

fpr, tpr, thresholds = roc_curve(y_delay_test, best_delay_risk)
youden_idx = np.argmax(tpr - fpr)
youden_threshold = thresholds[youden_idx]

print(f"\nYouden Index optimal threshold : {youden_threshold:.4f}")
print(f"  Sensitivity (TPR) : {tpr[youden_idx]:.4f}")
print(f"  Specificity       : {1 - fpr[youden_idx]:.4f}")

prec, rec, pr_thresholds = precision_recall_curve(y_delay_test, best_delay_risk)
f1_scores  = 2 * prec * rec / (prec + rec + 1e-9)
f1_opt_idx = np.argmax(f1_scores)
f1_threshold = pr_thresholds[f1_opt_idx] if f1_opt_idx < len(pr_thresholds) else pr_thresholds[-1]

print(f"F1-optimal threshold           : {f1_threshold:.4f}")
print(f"  F1-score  : {f1_scores[f1_opt_idx]:.4f}")
print(f"  Precision : {prec[f1_opt_idx]:.4f}")
print(f"  Recall    : {rec[f1_opt_idx]:.4f}")

# ── Decision zone logic (Youden Index based) ──────────────────────
#
#   HIGH_THRESHOLD = Youden threshold + 0.15  (upper zone: high-confidence late)
#   LOW_THRESHOLD  = Youden threshold - 0.15  (lower zone: high-confidence on-time)
#   Middle zone: "Monitor Closely"
#
# The +/- 0.15 offset is justified by the flat slope of the ROC curve
# around the threshold: the flat section represents uncertainty,
# so orders in that range warrant monitoring rather than a firm decision.

HIGH_THRESHOLD = min(youden_threshold + 0.15, 0.95)
LOW_THRESHOLD  = max(youden_threshold - 0.15, 0.05)

print(f"\nDelay decision zones (Youden Index based):")
print(f"  > {HIGH_THRESHOLD:.2f} -> Expedite Shipment   (high risk)")
print(f"  {LOW_THRESHOLD:.2f} - {HIGH_THRESHOLD:.2f} -> Monitor Closely     (medium risk)")
print(f"  < {LOW_THRESHOLD:.2f} -> No Action            (low risk)")

# ── Visualisation: both threshold methods ─────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax = axes[0]
fpr_all, tpr_all, _ = roc_curve(y_delay_test, best_delay_risk)
ax.plot(fpr_all, tpr_all, lw=2,
        label=f"{best_delay_name} (AUC={roc_auc_score(y_delay_test, best_delay_risk):.4f})")
ax.scatter(fpr[youden_idx], tpr[youden_idx], color="red", zorder=5,
           label=f"Youden point (thr={youden_threshold:.3f})")
ax.plot([0,1],[0,1],"k--",lw=1)
ax.set_xlabel("False Positive Rate"); ax.set_ylabel("True Positive Rate")
ax.set_title("ROC Curve — Youden Index threshold"); ax.legend(); ax.grid(alpha=0.3)

ax = axes[1]
ax.plot(rec, prec, lw=2, color="darkorange", label=f"{best_delay_name}")
ax.scatter(rec[f1_opt_idx], prec[f1_opt_idx], color="red", zorder=5,
           label=f"F1-optimum (thr={f1_threshold:.3f})")
ax.set_xlabel("Recall"); ax.set_ylabel("Precision")
ax.set_title("Precision-Recall Curve — F1-optimal threshold")
ax.legend(); ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ── Delay decision — Youden threshold applied ──────────────────────
delay_results = X_delay_test.copy()
delay_results["delay_risk"] = best_delay_risk

def delay_decision(row):
    if row["delay_risk"] > HIGH_THRESHOLD:
        return "Expedite Shipment"
    elif row["delay_risk"] > LOW_THRESHOLD:
        return "Monitor Closely"
    else:
        return "No Action"

delay_results["delay_decision"] = delay_results.apply(delay_decision, axis=1)

print("\n=== Delay Decision Distribution (Youden-based thresholds) ===")
print(delay_results["delay_decision"].value_counts())
```
